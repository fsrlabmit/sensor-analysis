For generation of the editing window of PE, for now, take [5:-5] of the sensor sequence and adjust it based on the indel size.  

```python
sensor_len = 55
```

```python
EA_LIB['indel_size'].value_counts()
```

```python
EA_LIB['wt_qwc'] = '5-50'
```

```python
def get_edit_wc(row):
    ref_seq_m = row['ref_seq_m_ori']
    alt_seq_m = row['alt_seq_m_ori']
    variant_type = row['Variant_Type']
    indel_size = row['indel_size']
    if (indel_size == 0) or (len(ref_seq_m) == len(alt_seq_m)):
        edit_qwc = '5-50'
    else:
        end_coor = 50 - len(ref_seq_m) + len(alt_seq_m)
        edit_qwc = f'5-{str(end_coor)}'

    return edit_qwc
```

```python
get_edit_wc(EA_LIB.iloc[0,:])
```

```python
EA_LIB[EA_LIB['Variant_Type']=='INS'][['indel_size','HGVSc_h']]
```

```python
EA_LIB['edit_qwc'] = EA_LIB.apply(get_edit_wc, axis=1)
EA_LIB['edit_qwc'].value_counts()
```
